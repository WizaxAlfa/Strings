```python
S[0]
```

# The second item from the left

```python
S[1]
```

# The last item from the end in S

```python
S[-1]
```

# The second-to-last item from the end

```python
S[-2]
```

# Negative indexing, the hard way

```python
S[len(S)-1]
```
